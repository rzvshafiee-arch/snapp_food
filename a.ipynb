```python
from datasets import load_dataset
import pandas as pd 
from sklearn.svm import LinearSVC
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import confusion_matrix,classification_report
```

# ngram_range=(1,8),min_df=3,max_df=0.98


```python
data = pd.read_csv("snapfood_train.csv")

x=data["comment"]
y=data["label"]

tf = TfidfVectorizer(ngram_range=(1,8),min_df=3,max_df=0.98,sublinear_tf=True)

xx= tf.fit_transform(x)

len(xx[8].toarray()[0])

model = LinearSVC()
model.fit(xx,y)

d_test = pd.read_csv("snapfood_test.csv")

o = []

for i in range(9033):
    p = model.predict(tf.transform([d_test["comment"].iloc[i]]))
    o.append(p[0])
o

t_data=pd.read_excel("test.xlsx")

t_data["prediction"]=o
t_data.to_excel("test.xlsx",index=False)

predictions=t_data["prediction"].values
truth=t_data["label"].values


print(classification_report(truth,predictions))
```

```python

import pandas as pd
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report


# 1. خواندن داده‌های آموزش
data = pd.read_csv("snapfood_train.csv")

x = data["comment"].fillna("").astype(str)
y = data["label"]


# 2. پاک‌سازی متن فارسی
def clean_text(text):
    text = text.replace("ي", "ی").replace("ك", "ک")
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"[^\u0600-\u06FF\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text


x = x.apply(clean_text)


# 3. تبدیل متن به ویژگی‌های عددی
tf = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

xx = tf.fit_transform(x)


# 4. آموزش مدل
model = LinearSVC(C=1.0)
model.fit(xx, y)


# 5. خواندن داده‌های آزمون
d_test = pd.read_csv("snapfood_test.csv")

test_comments = (
    d_test["comment"]
    .fillna("")
    .astype(str)
    .apply(clean_text)
)

# پیش‌بینی همه نظرات به صورت یک‌جا
predictions = model.predict(tf.transform(test_comments))

print(predictions)


# 6. ارزیابی مدل با برچسب‌های واقعی
t_data = pd.read_excel("test.xlsx")

if len(t_data) != len(predictions):
    raise ValueError(
        "تعداد ردیف‌های test.xlsx با پیش‌بینی‌ها برابر نیست."
    )

t_data["prediction"] = predictions
t_data.to_excel("test.xlsx", index=False)

truth = t_data["label"].values

print(classification_report(truth, predictions))
```
